```python
# [CHALLENGE TASK 1]

# CONSTANTS & INPUTS
LAST_NAME = "BERNAL"
SEED_NUM = 0  # Last digit of your student ID
FAVORITE_ARTIST = "BTS"


# 4. DECORATOR FOR DIAGNOSTIC LOGGING
def diagnostic_logger(func):
    def wrapper(*args, **kwargs):
        print(f"\n--- [LOG] Executing: {func.__name__} ---")
        result = func(*args, **kwargs)
        print(f"--- [LOG] Finished: {func.__name__} ---\n")
        return result
    return wrapper


# 1. GENERATE READINGS
def generate_readings(surname, seed, artist):
    # Generates a list of readings dynamically using your personalized variables
    raw_readings = [
        len(surname) * 10,
        seed * 15,
        len(artist) * 8,
        -5,            # Invalid reading for testing exception handling
        "ERROR",       # Invalid type for testing exception handling
        (seed + 5) * 12
    ]
    return raw_readings

# 2 & 3. VALIDATION WITH EXCEPTION HANDLING
def validate_readings(readings):
    valid_readings = []
    invalid_count = 0

    for item in readings:
        try:
            # Convert to float to verify numerical type
            val = float(item)
            if val < 0:
                raise ValueError("Reading cannot be negative.")
            valid_readings.append(val)
        except (ValueError, TypeError) as e:
            invalid_count += 1
            print(f"Validation Warning: Skipped invalid entry '{item}' -> Reason: {e}")

    return valid_readings, invalid_count

# CALCULATION & CLASSIFICATION FUNCTIONS
def calculate_metrics(*valid_data):
    if not valid_data:
        return 0, 0
    total = sum(valid_data)
    average = total / len(valid_data)
    return total, round(average, 2)

def classify_condition(avg_reading):
    if avg_reading > 50:
        return "CRITICAL OVERHEAT"
    elif avg_reading > 20:
        return "NORMAL OPERATION"
    else:
        return "LOW OUTPUT / STANDBY"


# 5. MAIN DIAGNOSTIC WORKFLOW & SUMMARY
@diagnostic_logger
def run_equipment_diagnostics():
    print(f"Diagnostic System Initialized for Student: {LAST_NAME}")
    print(f"Seed: {SEED_NUM} | Artist: {FAVORITE_ARTIST}")
    
    # Generate
    raw_data = generate_readings(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)
    print(f"Raw Generated Readings: {raw_data}")

    # Validate
    clean_data, error_count = validate_readings(raw_data)
    
    # Calculate
    total_val, avg_val = calculate_metrics(*clean_data)
    
    # Classify
    status = classify_condition(avg_val)

    # Display Summary
    print("==========================================")
    print("        EQUIPMENT DIAGNOSTIC SUMMARY       ")
    print("==========================================")
    print(f"Valid Readings Processed : {clean_data}")
    print(f"Invalid Entries Handled  : {error_count}")
    print(f"Total Reading Value      : {total_val}")
    print(f"Average Reading Value    : {avg_val}")
    print(f"Equipment Condition      : {status}")
    print("==========================================")

# Run the program
run_equipment_diagnostics()
```

```python
# [CHALLENGE TASK 2]
# ==========================================
# CONSTANTS & INPUTS
# ==========================================
LAST_NAME = "BERNAL"
SEED_NUM = 0
FAVORITE_ARTIST = "BTS"

# 1. GENERATE FAULT CODE
fault_code = f"{LAST_NAME}{SEED_NUM}{FAVORITE_ARTIST}"

# Global counter for recursive calls
call_count = 0

# 2 & 3. RECURSIVE FUNCTION WITH BASE CONDITION
def process_fault_trace(text):
    global call_count
    call_count += 1
    
    # Base condition (Requirement 3)
    if len(text) <= 1:
        print(f"  [Call #{call_count}] Base condition reached: '{text}'")
        return text

    # Recursive step logging
    print(f"  [Call #{call_count}] Processing '{text}' -> taking last character '{text[-1]}'")
    
    # Recursive call (Requirement 2)
    return text[-1] + process_fault_trace(text[:-1])

# ==========================================
# EXECUTION & DISPLAY
# ==========================================
print("==========================================")
print("       FAULT TRACE DIAGNOSTIC REPORT      ")
print("==========================================")
print(f"Generated Fault Data : {fault_code}")
print("\nRecursive Trace:")

# Process the fault code
call_count = 0  # Reset counter before invocation
final_result = process_fault_trace(fault_code)

# 4 & 5. ASSESSMENT DATA OUTPUTS
print("\n------------------------------------------")
print(f"Number of Recursive Calls : {call_count}")
print(f"Execution Log             : Processed {len(fault_code)} characters recursively.")
print(f"Final Output              : {final_result}")
print("==========================================")
```

```python
# [CHALLENGE TASK 3]

# main.py

# ==========================================
# CONSTANTS
# ==========================================
LAST_NAME = "BERNAL"
SEED_NUM = 0
FAVORITE_ARTIST = "BTS"

# ==========================================
# 1. DECORATOR FOR MONITORING
# ==========================================
def monitor_pipeline(func):
    def wrapper(*args, **kwargs):
        print(f"\n[PIPELINE MONITOR] Starting execution of: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"[PIPELINE MONITOR] Execution completed for: {func.__name__}\n")
        return result
    return wrapper

# ==========================================
# 2. GENERATOR FOR STREAMING DATA
# ==========================================
def telemetry_stream(surname, seed_num, artist):
    base_val = len(surname) + len(artist)
    
    raw_values = [
        base_val * 5,          # Normal reading
        (seed_num + 1) * 12,   # Normal reading
        "BAD_SENSOR",          # Invalid type for exception handling
        -15.0,                 # Out-of-bounds invalid reading
        base_val * 8,          # Abnormal elevated reading (triggers recursion)
        (seed_num + 2) * 10    # Normal reading
    ]
    
    for val in raw_values:
        yield val

# ==========================================
# 3. RECURSION FOR ABNORMAL CONDITION TRACE
# ==========================================
def trace_abnormal_condition(level, depth=1):
    print(f"  --> [Recursion Step {depth}] Analyzing abnormal level: {round(level, 2)}")
    
    if level <= 10:
        print(f"  --> Base condition reached at level {round(level, 2)}.")
        return depth
        
    return trace_abnormal_condition(level / 1.5, depth + 1)

# ==========================================
# MAIN PIPELINE WORKFLOW
# ==========================================
@monitor_pipeline
def run_monitoring_pipeline():
    processed_count = 0
    valid_readings = []
    invalid_count = 0
    abnormal_count = 0
    
    stream = telemetry_stream(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)
    
    print("--- REAL-TIME TELEMETRY STREAM PROCESSING ---")
    
    for raw_data in stream:
        processed_count += 1
        
        # Exception Handling
        try:
            val = float(raw_data)
            if val < 0:
                raise ValueError("Negative telemetry value detected.")
        except (ValueError, TypeError) as e:
            invalid_count += 1
            print(f"Skipped invalid value '{raw_data}': {e}")
            continue
            
        valid_readings.append(val)
        
        # Abnormal threshold check & recursion
        if val > 50:
            abnormal_count += 1
            print(f"\n[ALERT] Abnormal telemetry detected ({val})! Initiating recursive diagnostic trace...")
            steps = trace_abnormal_condition(val)
            print(f"[ALERT] Diagnostic trace completed in {steps} recursive steps.\n")

    # Lambda transformation
    transform_data = lambda readings: [round(x / 2, 2) for x in readings]
    transformed_readings = transform_data(valid_readings)

    # Status classification
    if abnormal_count > 0:
        overall_status = "WARNING - ATTENTION REQUIRED"
    elif invalid_count > len(valid_readings):
        overall_status = "CRITICAL - SENSOR FAILURE"
    else:
        overall_status = "OPTIMAL / STABLE"

    # Diagnostic Report Output
    print("==========================================")
    print("     INTELLIGENT MONITORING REPORT        ")
    print("==========================================")
    print(f"Student Metadata          : {LAST_NAME} | Seed: {SEED_NUM} | Artist: {FAVORITE_ARTIST}")
    print(f"Total Processed Readings  : {processed_count}")
    print(f"Valid Readings Count      : {len(valid_readings)}")
    print(f"Invalid Readings Count    : {invalid_count}")
    print(f"Abnormal Conditions Found : {abnormal_count}")
    print(f"Raw Valid Telemetry       : {valid_readings}")
    print(f"Transformed Telemetry     : {transformed_readings}")
    print(f"Overall Equipment Status  : {overall_status}")
    print("==========================================")

if __name__ == "__main__":
    run_monitoring_pipeline()
```
